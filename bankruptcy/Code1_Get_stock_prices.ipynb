```python
import pandas as pd
import numpy as np
```

```python
# Danh sách mã chứng khoán cần lấy lịch sử giá
symbols = ['PTI','PTS','PV2','PVB','PVC','PVE','PVG','PVI']
start_date = '2021-01-01'
end_date = '2023-12-31'

# Hàm lấy dữ liệu lịch sử giá
def fetch_stock_history(symbols, start, end):
    all_data = []
    for symbol in symbols:
        try:
            quote = Quote(symbol)
            data = quote.history(start=start, end=end, interval='1D')
            data['symbol'] = symbol  # Gắn thêm mã chứng khoán để dễ quản lý
            all_data.append(data)
        except Exception as e:
            print(f"Error fetching data for {symbol}: {e}")
    return pd.concat(all_data, ignore_index=True)
all_data = fetch_stock_history(symbols, start_date, end_date)
```

```python
all_data
```

```python
# lấy giá cổ phiếu
close_data = all_data[['time', 'symbol', 'close']]
pivot_data = close_data.pivot(index='time', columns='symbol', values='close')
stock = pivot_data.reset_index('time')
stock
```

```python
# lấy số lượng cổ phiếu
volume_data = all_data[['time', 'symbol', 'volume']]
pivot_data = volume_data.pivot(index='time', columns='symbol', values='volume')
stock11 = pivot_data.reset_index('time')
stock11
```

```python
stock['time'] = pd.to_datetime(stock['time'])
year_end_data = stock.groupby(stock['time'].dt.year).last()
year_end_prices = year_end_data[['time','PTI','PTS','PV2','PVB','PVC','PVE','PVG','PVI']]
year_end_prices
```

```python
stock11['time'] = pd.to_datetime(stock11['time'])
year_end_data11 = stock11.groupby(stock11['time'].dt.year).last()
year_end_volumes = year_end_data11[['time','PTI','PTS','PV2','PVB','PVC','PVE','PVG','PVI']]
print(year_end_volumes)
```
